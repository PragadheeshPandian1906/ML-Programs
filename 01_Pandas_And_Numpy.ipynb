```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# PART A: NUMPY
# ============================================================

print("\n========== NUMPY OPERATIONS ==========\n")

# 1. Creating arrays
a = np.array([10, 20, 30, 40, 50])
b = np.array([[1, 2, 3], [4, 5, 6]])

print("1D Array:", a)
print("2D Array:\n", b)
print("Zeros:\n", np.zeros((2, 3)))
print("Ones:\n", np.ones((2, 3)))
print("Full Array:\n", np.full((2, 2), 7))
print("Range:", np.arange(1, 11, 2))
print("Evenly spaced:", np.linspace(0, 1, 5))
print("Identity Matrix:\n", np.eye(3))
print("Array from existing shape:\n", np.zeros_like(b))

# 2. Array attributes
print("\n--- Array Attributes ---")
print("Shape:", b.shape)
print("Dimensions:", b.ndim)
print("Size:", b.size)
print("Data type:", b.dtype)
print("Item size:", b.itemsize)

# 3. Indexing and slicing
print("\n--- Indexing and Slicing ---")
print("First element:", a[0])
print("Last element:", a[-1])
print("Elements from index 1 to 3:", a[1:4])
print("Every second element:", a[::2])
print("First row:", b[0, :])
print("Second column:", b[:, 1])
print("Element at row 1, column 2:", b[1, 2])

# 4. Reshaping and dimensions
print("\n--- Reshaping ---")
x = np.arange(1, 13)
print("Original:", x)
print("Reshaped:\n", x.reshape(3, 4))
print("Flatten:", b.flatten())
print("Ravel:", b.ravel())
print("Transpose:\n", b.T)
print("Add dimension:", a[np.newaxis, :].shape)
print("Squeeze:", np.squeeze(np.array([[[10, 20, 30]]])))

# 5. Array operations
print("\n--- Arithmetic Operations ---")
print("Addition:", a + 5)
print("Subtraction:", a - 5)
print("Multiplication:", a * 2)
print("Division:", a / 10)
print("Power:", a ** 2)
print("Array addition:", np.array([1, 2, 3]) + np.array([4, 5, 6]))
print("Broadcasting:\n", b + np.array([10, 20, 30]))

# 6. Universal mathematical functions
print("\n--- Mathematical Functions ---")
print("Square root:", np.sqrt(a))
print("Exponential:", np.exp(np.array([1, 2, 3])))
print("Natural log:", np.log(a))
print("Absolute:", np.abs(np.array([-5, -10, 15])))
print("Sine:", np.sin(np.array([0, np.pi / 2, np.pi])))
print("Rounded:", np.round(np.array([1.234, 2.567]), 2))
print("Floor:", np.floor([1.8, 2.9, 3.1]))
print("Ceiling:", np.ceil([1.2, 2.1, 3.8]))

# 7. Aggregate and statistical functions
print("\n--- Statistics ---")
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Minimum:", np.min(a))
print("Maximum:", np.max(a))
print("Standard deviation:", np.std(a))
print("Variance:", np.var(a))
print("Percentiles:", np.percentile(a, [25, 50, 75]))
print("Cumulative sum:", np.cumsum(a))
print("Cumulative product:", np.cumprod(np.array([1, 2, 3, 4])))
print("Sum by column:", np.sum(b, axis=0))
print("Sum by row:", np.sum(b, axis=1))
print("Argmin:", np.argmin(a))
print("Argmax:", np.argmax(a))

# 8. Boolean indexing and filtering
print("\n--- Filtering ---")
print("Values greater than 25:", a[a > 25])
print("Values divisible by 20:", a[a % 20 == 0])
print("Any value > 45:", np.any(a > 45))
print("All values > 0:", np.all(a > 0))
print("Count values > 25:", np.count_nonzero(a > 25))
print("Where condition:", np.where(a > 25, "High", "Low"))
print("Clip values:", np.clip(a, 20, 40))

# 9. Sorting, searching and unique values
print("\n--- Sorting and Searching ---")
c = np.array([30, 10, 20, 10, 40])
print("Sorted:", np.sort(c))
print("Sorted indices:", np.argsort(c))
print("Unique values:", np.unique(c))
print("Search index:", np.where(c == 20))
print("Indices of nonzero:", np.nonzero(np.array([0, 1, 0, 3])))
print("Membership:", np.isin(c, [10, 40]))

# 10. Joining, splitting and stacking
print("\n--- Array Combining ---")
p = np.array([1, 2, 3])
q = np.array([4, 5, 6])
print("Concatenate:", np.concatenate((p, q)))
print("Vertical stack:\n", np.vstack((p, q)))
print("Horizontal stack:", np.hstack((p, q)))
print("Column stack:\n", np.column_stack((p, q)))
print("Split:", np.split(np.arange(8), 4))

# 11. Random number generation
print("\n--- Random Numbers ---")
rng = np.random.default_rng(42)
print("Random floats:", rng.random(5))
print("Random integers:", rng.integers(1, 100, size=5))
print("Normal distribution:", rng.normal(0, 1, size=5))
print("Random choice:", rng.choice([10, 20, 30], size=5))
print("Random matrix:\n", rng.integers(1, 10, size=(2, 3)))

# 12. Set operations
print("\n--- Set Operations ---")
u = np.array([1, 2, 3, 4])
v = np.array([3, 4, 5, 6])
print("Union:", np.union1d(u, v))
print("Intersection:", np.intersect1d(u, v))
print("Difference:", np.setdiff1d(u, v))
print("Symmetric difference:", np.setxor1d(u, v))

# 13. Linear algebra
print("\n--- Linear Algebra ---")
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])
print("Matrix multiplication:\n", np.matmul(m1, m2))
print("Dot product:", np.dot(np.array([1, 2]), np.array([3, 4])))
print("Determinant:", np.linalg.det(m1))
print("Inverse:\n", np.linalg.inv(m1))
print("Eigenvalues:", np.linalg.eigvals(m1))
print("Solve equations:", np.linalg.solve(
    np.array([[2, 1], [1, 3]]),
    np.array([5, 6])
))

# 14. Missing and special values
print("\n--- Special Values ---")
missing = np.array([10.0, 20.0, np.nan, 40.0])
print("Is NaN:", np.isnan(missing))
print("NaN count:", np.isnan(missing).sum())
print("Mean ignoring NaN:", np.nanmean(missing))
print("Infinity check:", np.isinf([1, np.inf, -np.inf]))
print("Finite check:", np.isfinite([1, np.inf, np.nan]))


# ============================================================
# PART B: PANDAS
# ============================================================

print("\n========== PANDAS OPERATIONS ==========\n")

# 15. Creating a Series
s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])
print("Series:\n", s)
print("Values:", s.values)
print("Index:", s.index)
print("Data type:", s.dtype)
print("Element by label:", s.loc["b"])
print("Element by position:", s.iloc[1])
print("Series mean:", s.mean())
print("Series unique:", pd.Series([1, 2, 2, 3]).unique())
print("Value counts:\n", pd.Series(["A", "B", "A"]).value_counts())

# 16. Creating a DataFrame
df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Charan", "Divya", "Esha", "Farhan"],
    "Age": [20, 21, 19, 22, 20, 21],
    "Department": ["IT", "CSE", "IT", "ECE", "CSE", "IT"],
    "Marks": [85, 92, 78, 88, 95, 70],
    "Attendance": [90, 85, 95, 80, 92, 75]
})

print("\nDataFrame:\n", df)

# 17. Inspecting data
print("\n--- Inspection ---")
print("First 3 rows:\n", df.head(3))
print("Last 2 rows:\n", df.tail(2))
print("Shape:", df.shape)
print("Columns:", df.columns)
print("Index:", df.index)
print("Data types:\n", df.dtypes)
print("Information:")
df.info()
print("Descriptive statistics:\n", df.describe())
print("Number of rows:", len(df))
print("Total elements:", df.size)
print("Column names:", df.columns.tolist())
print("Index of rows:", df.index.tolist())

# 18. Selecting rows and columns
print("\n--- Selection ---")
print("Single column:\n", df["Name"])
print("Multiple columns:\n", df[["Name", "Marks"]])
print("First row by position:\n", df.iloc[0])
print("First two rows:\n", df.iloc[:2])
print("Rows by label:\n", df.loc[0:2, ["Name", "Marks"]])
print("Single cell:", df.loc[0, "Marks"])
print("Cell by position:", df.iloc[0, 3])

# 19. Filtering data
print("\n--- Filtering ---")
print("Marks > 80:\n", df[df["Marks"] > 80])
print("Marks > 80 and Age >= 20:\n",
      df[(df["Marks"] > 80) & (df["Age"] >= 20)])
print("Department is IT:\n", df[df["Department"] == "IT"])
print("Department in IT or CSE:\n",
      df[df["Department"].isin(["IT", "CSE"])])
print("Name starts with A:\n", df[df["Name"].str.startswith("A")])
print("Marks between 80 and 90:\n",
      df[df["Marks"].between(80, 90)])
print("Query method:\n", df.query("Marks >= 85 and Age >= 20"))

# 20. Adding, modifying and deleting columns/rows
print("\n--- Modification ---")
df["Result"] = np.where(df["Marks"] >= 40, "Pass", "Fail")
df["Bonus"] = df["Marks"] + 5
print("Added columns:\n", df)

df["Marks"] = df["Marks"].clip(0, 100)
df.loc[0, "Marks"] = 87
print("Updated DataFrame:\n", df)

print("Drop Bonus column:\n", df.drop(columns=["Bonus"]))
print("Drop row with index 0:\n", df.drop(index=0))

# 21. Sorting
print("\n--- Sorting ---")
print("Sort by Marks:\n", df.sort_values("Marks"))
print("Descending order:\n", df.sort_values("Marks", ascending=False))
print("Sort by Department then Marks:\n",
      df.sort_values(["Department", "Marks"],
                     ascending=[True, False]))
print("Sort by index:\n", df.sort_index())

# 22. Missing values
print("\n--- Missing Values ---")
dirty = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Marks": [80, np.nan, 90, np.nan],
    "Age": [20, 21, np.nan, 22]
})

print("Data with missing values:\n", dirty)
print("Missing-value mask:\n", dirty.isna())
print("Missing count by column:\n", dirty.isna().sum())
print("Non-missing count:\n", dirty.notna().sum())
print("Drop rows with missing values:\n", dirty.dropna())
print("Fill all missing values:\n", dirty.fillna(0))
print("Fill Marks with mean:\n",
      dirty.assign(Marks=dirty["Marks"].fillna(dirty["Marks"].mean())))
print("Forward fill:\n", dirty.ffill())
print("Backward fill:\n", dirty.bfill())

# 23. Duplicate handling
print("\n--- Duplicates ---")
duplicates = pd.DataFrame({
    "Name": ["A", "B", "A", "C"],
    "Marks": [80, 90, 80, 75]
})
print("Duplicate mask:\n", duplicates.duplicated())
print("Remove duplicate rows:\n", duplicates.drop_duplicates())
print("Duplicates by Name:\n",
      duplicates.duplicated(subset=["Name"]))

# 24. String functions
print("\n--- String Operations ---")
names = pd.Series(["  arun kumar ", "BALA", "charan", "divya"])
print("Lowercase:\n", names.str.lower())
print("Uppercase:\n", names.str.upper())
print("Title case:\n", names.str.title())
print("Remove spaces:\n", names.str.strip())
print("Contains 'a':\n", names.str.contains("a", case=False))
print("Replace text:\n", names.str.replace("a", "@", regex=False))
print("String length:\n", names.str.len())
print("Split names:\n", names.str.strip().str.split())

# 25. Apply, map and replace
print("\n--- Apply, Map and Replace ---")
print("Map departments:\n",
      df["Department"].map({"IT": "Information Technology",
                            "CSE": "Computer Science",
                            "ECE": "Electronics"}))
print("Apply function:\n", df["Marks"].apply(lambda x: x + 2))
print("Replace values:\n",
      df["Department"].replace({"IT": "Information Technology"}))
print("Conditional column:\n",
      df.assign(Grade=df["Marks"].apply(
          lambda x: "A" if x >= 90 else "B" if x >= 80 else "C"
      )))

# 26. GroupBy and aggregation
print("\n--- GroupBy and Aggregation ---")
print("Average marks by department:\n",
      df.groupby("Department")["Marks"].mean())
print("Multiple statistics:\n",
      df.groupby("Department")["Marks"].agg(["mean", "max", "min", "count"]))
print("Multiple columns:\n",
      df.groupby("Department").agg({
          "Marks": ["mean", "max"],
          "Age": "mean"
      }))
print("Group sizes:\n", df.groupby("Department").size())
print("Value counts:\n", df["Department"].value_counts())
print("Crosstab:\n", pd.crosstab(df["Department"], df["Result"]))

# 27. Combining DataFrames
print("\n--- Concatenation, Merge and Join ---")
df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Arun", "Bala", "Charan"]
})
df2 = pd.DataFrame({
    "ID": [1, 2, 4],
    "Marks": [80, 90, 85]
})

print("Concatenate rows:\n", pd.concat([df1, df1], ignore_index=True))
print("Inner merge:\n", pd.merge(df1, df2, on="ID", how="inner"))
print("Left merge:\n", pd.merge(df1, df2, on="ID", how="left"))
print("Right merge:\n", pd.merge(df1, df2, on="ID", how="right"))
print("Outer merge:\n", pd.merge(df1, df2, on="ID", how="outer"))

left = df1.set_index("ID")
right = df2.set_index("ID")
print("Join by index:\n", left.join(right, how="left"))

# 28. Reshaping data
print("\n--- Reshaping ---")
wide = pd.DataFrame({
    "Name": ["Arun", "Bala"],
    "Maths": [80, 90],
    "Science": [85, 95]
})
print("Wide format:\n", wide)
long = wide.melt(
    id_vars="Name",
    var_name="Subject",
    value_name="Marks"
)
print("Long format using melt:\n", long)
print("Pivot back to wide:\n",
      long.pivot(index="Name", columns="Subject", values="Marks"))
print("Transpose:\n", wide.T)

# 29. Date and time functions
print("\n--- Date and Time ---")
dates = pd.to_datetime(["2026-01-10", "2026-02-15", "2026-03-20"])
date_df = pd.DataFrame({"Date": dates, "Sales": [100, 200, 300]})
print(date_df)
print("Year:", date_df["Date"].dt.year.tolist())
print("Month:", date_df["Date"].dt.month.tolist())
print("Day:", date_df["Date"].dt.day.tolist())
print("Day of week:", date_df["Date"].dt.day_name().tolist())
print("Date range:", pd.date_range("2026-01-01", periods=5, freq="D"))

# 30. Statistical functions and correlation
print("\n--- Data Analysis ---")
print("Mean marks:", df["Marks"].mean())
print("Median marks:", df["Marks"].median())
print("Standard deviation:", df["Marks"].std())
print("Variance:", df["Marks"].var())
print("Minimum:", df["Marks"].min())
print("Maximum:", df["Marks"].max())
print("Quantiles:\n", df["Marks"].quantile([0.25, 0.50, 0.75]))
print("Correlation matrix:\n",
      df[["Age", "Marks", "Attendance"]].corr())
print("Covariance matrix:\n",
      df[["Age", "Marks", "Attendance"]].cov())
print("Rank of marks:\n", df["Marks"].rank(ascending=False))
print("Cumulative marks:\n", df["Marks"].cumsum())
print("Difference between rows:\n", df["Marks"].diff())

# 31. CSV file operations
print("\n--- CSV Operations ---")
df.to_csv("students.csv", index=False)
loaded_df = pd.read_csv("students.csv")
print("Data loaded from CSV:\n", loaded_df)
print("CSV file created successfully.")

# Other useful file functions:
# pd.read_excel("students.xlsx")
# df.to_excel("students.xlsx", index=False)
# pd.read_json("students.json")
# df.to_json("students.json", orient="records", indent=2)

# 32. Data visualization
print("\n--- Visualization ---")

# Bar chart
df.plot(x="Name", y="Marks", kind="bar", legend=False)
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.tight_layout()
plt.show()

# Histogram
df["Marks"].plot(kind="hist", bins=5, edgecolor="black")
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.tight_layout()
plt.show()

# Scatter plot
df.plot(x="Attendance", y="Marks", kind="scatter")
plt.title("Attendance vs Marks")
plt.tight_layout()
plt.show()

# NumPy line plot
x_values = np.arange(1, 11)
y_values = x_values ** 2
plt.plot(x_values, y_values, marker="o")
plt.title("NumPy: Square of Numbers")
plt.xlabel("x")
plt.ylabel("x squared")
plt.grid(True)
plt.tight_layout()
plt.show()

print("\n========== PROGRAM COMPLETED ==========")
```
